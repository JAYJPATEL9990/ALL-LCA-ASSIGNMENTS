```python
# Define two matrices
X = [[1, 2, 3], 
     [4, 5, 6], 
     [7, 8, 9]]

Y = [[9, 8, 7], 
     [6, 5, 4], 
     [3, 2, 1]]

# Initialize result matrix with zeros
result = [[0, 0, 0], 
          [0, 0, 0], 
          [0, 0, 0]]

# Perform addition using nested loops
for i in range(len(X)):
    for j in range(len(X[0])):
        result[i][j] = X[i][j] + Y[i][j]

# Print the result
for r in result:
    print(r)
```

```python
import numpy as np

# Define two 2x2 matrices
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])

# Method 1: Using the + operator
sum_operator = matrix_a + matrix_b

# Method 2: Using the np.add() function
sum_function = np.add(matrix_a, matrix_b)

print("Sum using + operator:")
print(sum_operator)

print("\nSum using np.add():")
print(sum_function)
                      
```
