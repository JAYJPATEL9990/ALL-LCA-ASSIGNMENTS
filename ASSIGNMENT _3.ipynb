```python

# Function to check right-angled triangle

def check_right_triangle(a, b, c):
    # Find the largest side
    largest = max(a, b, c)

    # Check Pythagoras theorem
    if largest == a:
        return b**2 + c**2 == a**2
    elif largest == b:
        return a**2 + c**2 == b**2
    else:
        return a**2 + b**2 == c**2


# Taking input from the user
a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

# Calling the function
if check_right_triangle(a, b, c):
    print("The triangle is a right-angled triangle.")
else:
    print("The triangle is not a right-angled triangle.")
```

```python

# Exercise Q1

def prime_factors(n):
    factors = []
    i = 2

    while i <= n:
        if n % i == 0:
            factors.append(i)
            n = n // i
        else:
            i += 1

    return factors


num = int(input("Enter a number: "))

print("Prime factors:", prime_factors(num))
```

```python

# Exercise Q3

import math

def quadratic_roots(a, b, c):
    # Calculate the discriminant
    d = b**2 - 4*a*c

    # If d is negative, real roots are not possible
    if d < 0:
        raise ValueError("math domain error")

    root1 = (-b + math.sqrt(d)) / (2*a)
    root2 = (-b - math.sqrt(d)) / (2*a)

    return root1, root2


try:
    a = float(input("Enter a: "))
    b = float(input("Enter b: "))
    c = float(input("Enter c: "))

    if a == 0:
        raise ValueError("a cannot be zero")

    roots = quadratic_roots(a, b, c)
    print("The roots are:", roots)

except ValueError as e:
    print("ValueError:", e)

except NameError as e:
    print("NameError:", e)

except TypeError as e:
    print("TypeError:", e)
```
